```python
def change_DPI(image_path, save_path, scale):
  image = Image.open(image_path)
  dpi = image.info.get('dpi')
  if (dpi == None or dpi == (0,0)):
    image.save(save_path, 'JPEG')

  else:
    mod_dpi  = tuple(int(i * scale) for i in dpi)
    image.save(save_path, 'JPEG', dpi=mod_dpi)
```

```python
def compress(image_path, save_path, quality):
  image = Image.open(image_path)
  image.save(save_path, 'JPEG', quality=quality)
```

```python
def resize(image_path, save_path, scale):
  image = Image.open(image_path)
  w, h = image.size
  new_w = int(w * scale)
  new_h = int(h * scale)
  image = image.resize((new_w, new_h))
  image.save(save_path, 'JPEG')
```

```python
def res_change(dataset, save_path, scale):
  for root, dirs, files in os.walk(dataset):
        # Create the corresponding structure in the output directory
        rel_path = os.path.relpath(root, dataset)
        save_dir = os.path.join(save_path, rel_path)
        if not os.path.exists(save_dir):
            os.makedirs(save_dir)

        for file in files:
            if file.lower().endswith(('.png', '.jpg', '.jpeg')):
                input_path = os.path.join(root, file)
                parts = file.split('.')
                img_name = f'{parts[0]}_dpi{int(scale*100)}.{parts[1]}'
                output_path = os.path.join(save_dir, img_name)
                change_DPI(input_path, output_path, scale)
```

```python
def img_compress(dataset, save_path, quality):
  for root, dirs, files in os.walk(dataset):
        # Create the corresponding structure in the output directory
        rel_path = os.path.relpath(root, dataset)
        save_dir = os.path.join(save_path, rel_path)

        if not os.path.exists(save_dir):
            os.makedirs(save_dir)
            print(save_dir)

        for file in files:
            if file.lower().endswith(('.png', '.jpg', '.jpeg')):
                input_path = os.path.join(root, file)
                parts = file.split('.')
                img_name = f'{parts[0]}_qual{quality}.{parts[1]}'
                output_path = os.path.join(save_dir, img_name)
                compress(input_path, output_path, quality)
```

```python
def img_resize(dataset, save_path, scale):
  for root, dirs, files in os.walk(dataset):
        # Create the corresponding structure in the output directory
        rel_path = os.path.relpath(root, dataset)
        save_dir = os.path.join(save_path, rel_path)

        if not os.path.exists(save_dir):
            os.makedirs(save_dir)
            print(save_dir)

        for file in files:
            if file.lower().endswith(('.png', '.jpg', '.jpeg')):
                input_path = os.path.join(root, file)
                parts = file.split('.')
                img_name = f'{parts[0]}_size{scale*100}.{parts[1]}'
                output_path = os.path.join(save_dir, img_name)
                resize(input_path, output_path, scale)
```
